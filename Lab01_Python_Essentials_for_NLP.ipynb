```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 04 (a)
raw = ["nlp", "ai", "nlp", "deep", "learning", "ai", "nlp", "python"]
vocab = set(raw)

print("raw list              :", raw, "-> total items:", len(raw))
print("vocab set             :", vocab)
print("Number of unique words:", len(vocab))

# Why do sets remove duplicates automatically?
# A set stores its items in a hash table. Before inserting a word, Python computes
# hash(word) and checks that slot: if an equal item is already there, the insert is
# simply skipped instead of adding a second copy. So a set can only ever hold one
# instance of each distinct value, which is exactly what we want for a vocabulary.
```

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 06 (d)
def normalize(text):
    """Strip whitespace, lowercase, and drop commas and full-stops."""
    text = text.strip()
    text = text.lower()
    text = text.replace(",", "")
    text = text.replace(".", "")
    return text

messy_inputs = [
    "   Natural Language Processing, is FUN.   ",
    "  Deep Learning, Machine Learning, and AI.  ",
    "   TOKENIZATION. splits, text.   "
]

for raw_text in messy_inputs:
    print("Raw     :", repr(raw_text))
    print("Cleaned :", repr(normalize(raw_text)))
    print("-" * 55)
```

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 07 (c)
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words = []
pure_numbers = []
mixed_tokens = []

for w in words:
    if w.isalpha():
        pure_words.append(w)
    elif w.isdigit():
        pure_numbers.append(w)
    else:
        mixed_tokens.append(w)

print("Original list :", words)
print()
print("pure_words    :", pure_words)
print("pure_numbers  :", pure_numbers)
print("mixed_tokens  :", mixed_tokens)
```

### Task 07 (d)

> Write a function called `validate_token(token)` that returns the string "WORD" if the token is purely alphabetic, "NUMBER" if it is purely numeric, and "MIXED" otherwise. Apply this function to each element in the list `["chat", "99", "gpt4", "bert", "2024", "nlp!"]` using a loop and print the token alongside its label.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 07 (d)
def validate_token(token):
    """Classify a token as WORD, NUMBER or MIXED."""
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

test_tokens = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]

for t in test_tokens:
    print(f"{t:<6} -> {validate_token(t)}")
```

## Task 08: Functions and Lambda for NLP Pipelines

**Topic:** Functions, Lambda - Reusable Text Processing

### Task 08 (a)

> Write a regular function called `clean(text)` that takes a string, strips whitespace, converts it to lowercase, and returns the result. Then write an equivalent lambda function called `clean_lambda` that performs the same two operations in one line. Test both on the string "  HELLO Natural Language Processing  " and print their outputs.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 08 (a)
def clean(text):
    """Strip surrounding whitespace and lowercase the text."""
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

sample = "  HELLO Natural Language Processing  "

print("Input            :", repr(sample))
print("clean()          :", repr(clean(sample)))
print("clean_lambda()   :", repr(clean_lambda(sample)))
print("Both are equal?  :", clean(sample) == clean_lambda(sample))
```

### Task 08 (b)

> Write a function called `tokenize(text)` that takes a cleaned string and returns a list of individual word tokens using `split()`. Then write another function called `remove_short(tokens, n)` that takes the token list and an integer n, and returns only tokens whose length is greater than n. Chain both functions on the sentence "I love working with NLP tools in Python daily" with n = 3 and print the result.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 08 (b)
def tokenize(text):
    """Split a cleaned string into a list of word tokens."""
    return text.split()

def remove_short(tokens, n):
    """Keep only the tokens longer than n characters."""
    return [t for t in tokens if len(t) > n]

sentence_08b = "I love working with NLP tools in Python daily"

tokens_08b = tokenize(clean(sentence_08b))
print("Tokenised            :", tokens_08b)

result_08b = remove_short(tokens_08b, 3)
print("After remove_short(3):", result_08b)
```

### Task 08 (c)

> Write a function called `word_lengths(tokens)` that takes a list of tokens and returns a dictionary where each key is a word and each value is the length of that word. Call the function on `tokens = ["language", "model", "ai", "processing", "deep"]` and print the resulting dictionary.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 08 (c)
def word_lengths(tokens):
    """Map every token to its character length."""
    return {t: len(t) for t in tokens}

tokens_08c = ["language", "model", "ai", "processing", "deep"]

print("Tokens :", tokens_08c)
print("Lengths:", word_lengths(tokens_08c))
```

### Task 08 (d)

> Combine the functions from parts (b) and (c) into a full pipeline function called `nlp_pipeline(text, min_len)` that: cleans the text, tokenises it, removes short words, and returns a length dictionary. Call this pipeline on two different sentences and print the outputs.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 08 (d)
def nlp_pipeline(text, min_len):
    """clean -> tokenize -> remove_short -> word_lengths"""
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    kept = remove_short(tokens, min_len)
    return word_lengths(kept)

sentence_1 = "   Natural Language Processing is an Exciting FIELD   "
sentence_2 = "  Deep learning models can read and generate human text  "

print("Sentence 1:", repr(sentence_1))
print("Pipeline  :", nlp_pipeline(sentence_1, 3))
print()
print("Sentence 2:", repr(sentence_2))
print("Pipeline  :", nlp_pipeline(sentence_2, 4))
```

> Import `Counter` from `collections`. Given the sentence "nlp nlp python ai nlp python text ai nlp python text", split it into words and create a Counter called `freq`. Print the full counter, then print: (i) the 3 most common words using `most_common()`, (ii) the frequency of the word "python", and (iii) the total number of words and the size of the unique vocabulary.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 09 (d)
from collections import Counter

sentence_09d = "nlp nlp python ai nlp python text ai nlp python text"
words_09d = sentence_09d.split()

freq = Counter(words_09d)

print("Sentence    :", sentence_09d)
print("Full Counter:", freq)
print()
print("(i)   3 most common words :", freq.most_common(3))
print("(ii)  Frequency of 'python':", freq["python"])
print("(iii) Total number of words:", sum(freq.values()))
print("      Unique vocabulary size:", len(freq))
```
