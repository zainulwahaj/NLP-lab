## Task 01: Regex Fundamentals (Character Classes & Quantifiers)

**Topic:** Character classes and quantifiers on a text with words, numbers and symbols

### Task 01 (a)

> Extract all digits from the text.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 01 (a)
sample_text = ("On 14 March 2021, Dr. Ayesha Khan travelled from Lahore to Karachi and Quetta. "
               "She bought 3 laptops for Rs. 450000 in 2019, then 12 more in 2020! "
               "The CS-101 class has 45 students & 2 sections (see room B-7). "
               "Receipt 1234567 was issued at 9:30 for $12.50 + 5% tax.")

print(sample_text)

digits = re.findall(r"\d", sample_text)          # \d = one digit character
print("\nAll digits, one by one:", digits)
print("Number of digits:", len(digits))
```

### Task 01 (b)

> Extract all words that start with a capital letter.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 01 (b)
# \b = word boundary, [A-Z] = a capital letter, \w* = the rest of the word
capital_words = re.findall(r"\b[A-Z]\w*", sample_text)
print(capital_words)
print("Count:", len(capital_words))
print("\nNote: 'CS' (from CS-101) and 'B' (from B-7) are included because they also start with a capital.")
```

### Task 01 (c)

> Extract all sequences of exactly 4 digits (e.g., years).

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 01 (c)
# (?<!\d) and (?!\d) mean "not preceded / not followed by another digit",
# so 1234567 (7 digits) and 450000 (6 digits) are not cut into pieces of 4
four_digits = re.findall(r"(?<!\d)\d{4}(?!\d)", sample_text)
print(four_digits)

# What happens without the lookarounds: it wrongly pulls 4-digit chunks out of longer numbers
print("Naive \\d{4} pattern:", re.findall(r"\d{4}", sample_text))
```

### Task 01 (d)

> Extract all words that are between 3 to 6 characters long.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 01 (d)
# [A-Za-z]{3,6} with word boundaries on both sides -> whole words of 3 to 6 letters
mid_words = re.findall(r"\b[A-Za-z]{3,6}\b", sample_text)
print(mid_words)
print("Count:", len(mid_words))
print("\nLeft out: words that are too short (On, to, in, of) or too long (travelled, Karachi, students...).")
```

## Task 02: Pattern-Based Extraction (Real World Data)

**Topic:** Extracting emails, phone numbers, dates and percentages from a mixed paragraph

### Task 02 (a)

> Write a regex pattern using re.findall() to extract all email addresses.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 02 (a)
paragraph = ("Order ORD-2024-0017 was placed on 05/03/2024 by ali.raza@example.com, who can be reached "
             "on 0300-1234567 or 0321-7654321. A 15% discount and 2.5% cashback were applied. "
             "A second order, ORD-2024-0018, was booked on 17/11/2023 by sara_k99@mail.co.uk "
             "(phone 0333-5551234) with a 12.5% service charge. Please ignore 31/02/2024x, the number "
             "0415-1234567 and the text support@ because they are not valid.")

print(paragraph)

email_pattern = r"[\w.+-]+@[\w-]+(?:\.[\w-]+)+"       # name @ domain . tld (the domain may have several dots)
emails = re.findall(email_pattern, paragraph)
print("\nEmails:", emails)
```

### Task 02 (b)

> Write a regex pattern to extract all phone numbers (assume format: 03XX-XXXXXXX).

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 02 (b)
# 03 + two more digits, a hyphen, then exactly seven digits
phone_pattern = r"\b03\d{2}-\d{7}\b"
phones = re.findall(phone_pattern, paragraph)
print("Phone numbers:", phones)
print("Not matched on purpose: '0415-1234567' (does not start with 03).")
```

### Task 02 (c)

> Write a regex pattern to extract all dates in DD/MM/YYYY format.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 02 (c)
# day 01-31, month 01-12, any 4-digit year; the (?!\w) at the end rejects '31/02/2024x'
date_pattern = r"\b(?:0[1-9]|[12]\d|3[01])/(?:0[1-9]|1[0-2])/\d{4}(?!\w)"
dates = re.findall(date_pattern, paragraph)
print("Dates:", dates)
print("\nA pattern can only check the shape: 31/02/2024 would still match without the trailing 'x'")
print("even though 31 February does not exist. Real validation needs datetime.")
```

### Task 02 (d)

> Write a regex pattern to extract all percentage values (e.g., 45%, 12.5%).

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 02 (d)
percent_pattern = r"\b\d+(?:\.\d+)?%"            # digits, an optional .digits part, then %
percentages = re.findall(percent_pattern, paragraph)
print("Percentages:", percentages)

# Check every extraction against the values written into the paragraph
assert emails == ["ali.raza@example.com", "sara_k99@mail.co.uk"]
assert phones == ["0300-1234567", "0321-7654321", "0333-5551234"]
assert dates == ["05/03/2024", "17/11/2023"]
assert percentages == ["15%", "2.5%", "12.5%"]
print("\nAll four extractions match the expected values.")
```

## Task 03: Word Tokenization (NLTK vs spaCy)

**Topic:** Comparing NLTK word_tokenize() with a blank spaCy pipeline

### Task 03 (a)

> Tokenize the paragraph using NLTK's word_tokenize().

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 03 (a)
text = ("I don't think it's ready yet, but Dr. Smith won't wait. "
        "They're sending 3 e-mails (see www.example.com) to the U.S. office, and it can't be late. "
        'She said, "We\'ll finish by 5 p.m." Isn\'t that great? Mr. Khan paid $50 for state-of-the-art tools.')

print(text, "\n")

from nltk.tokenize import word_tokenize
nltk_tokens = word_tokenize(text)
print("NLTK tokens:")
print(nltk_tokens)
```

### Task 03 (b)

> Tokenize the same paragraph using a blank spaCy pipeline.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 03 (b)
nlp_blank = spacy.blank("en")                 # only the tokenizer, no trained model
spacy_doc = nlp_blank(text)
spacy_tokens = [t.text for t in spacy_doc]

print("spaCy tokens:")
print(spacy_tokens)
```

### Task 03 (c)

> Compare and write down (in a comment) at least 2 differences you noticed between the two outputs.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 03 (c)
# Let the code find the differences first, then describe them in the comment below.
only_nltk = [t for t in nltk_tokens if t not in spacy_tokens]
only_spacy = [t for t in spacy_tokens if t not in nltk_tokens]
print("Tokens only NLTK produced :", sorted(set(only_nltk)))
print("Tokens only spaCy produced:", sorted(set(only_spacy)))

# DIFFERENCES I NOTICED
# 1. Quotation marks: NLTK rewrites the straight quote " as `` (opening) and '' (closing),
#    while spaCy keeps the original " character as its own token.
# 2. Hyphenated words: NLTK keeps 'e-mails' and 'state-of-the-art' as single tokens, but spaCy
#    splits them at every hyphen (e / - / mails and state / - / of / - / the / - / art).
# 3. Both tokenizers agree on contractions (do + n't, wo + n't, 'll, 're), on abbreviations
#    such as Dr., U.S. and p.m., on the URL www.example.com and on "$50" -> "$" + "50".
# 4. Because of difference 2 spaCy gives more tokens (see the count below): 2 extra for
#    'e-mails' and 6 extra for 'state-of-the-art'.
```

### Task 03 (d)

> Count the total number of tokens produced by each method.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 03 (d)
print("NLTK  word_tokenize :", len(nltk_tokens), "tokens")
print("spaCy blank('en')   :", len(spacy_tokens), "tokens")
print("Difference          :", abs(len(nltk_tokens) - len(spacy_tokens)))
```

## Task 04: Token Attributes (Categorizing Tokens)

**Topic:** Token attributes: is_alpha, like_num, like_email, is_currency

### Task 04 (a)

> Load the sentence into a spaCy Doc object.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 04 (a)
sentence = "Ali paid $50 and Rs. 500 for 3 books, then emailed ali.khan@example.com on 12 May 2024."
doc = nlp_blank(sentence)

print("Sentence :", sentence)
print("Type     :", type(doc))
print("Tokens   :", len(doc))
print([token.text for token in doc])
```

### Task 04 (b)

> Loop through each token and print its is_alpha value.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 04 (b)
print(f"{'token':<22}{'is_alpha':>9}")
print("-" * 31)
for token in doc:
    print(f"{token.text:<22}{str(token.is_alpha):>9}")
```

### Task 04 (c)

> Print each token's like_num and like_email values.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 04 (c)
print(f"{'token':<22}{'like_num':>9}{'like_email':>12}")
print("-" * 43)
for token in doc:
    print(f"{token.text:<22}{str(token.like_num):>9}{str(token.like_email):>12}")
```

### Task 04 (d)

> Identify and list only the tokens where is_currency is True.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 04 (d)
currency_tokens = [token.text for token in doc if token.is_currency]
print("Tokens with is_currency == True:", currency_tokens)

# Check the other currency forms in the sentence too
for token in doc:
    if token.text in ("$", "Rs", "Rs."):
        print(f"   {token.text!r:<6} is_currency = {token.is_currency}")
print("\nOnly the symbol is recognised; spaCy does not treat the word 'Rs.' as a currency sign.")
```

## Task 05: Sentence Segmentation (Handling Tricky Cases)

**Topic:** Sentence segmentation with abbreviations

### Task 05 (a)

> Load a trained spaCy pipeline (en_core_web_sm).

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 05 (a)
nlp = spacy.load("en_core_web_sm")
print("Loaded pipeline :", nlp.meta["lang"] + "_" + nlp.meta["name"], nlp.meta["version"])
print("Components      :", nlp.pipe_names)
```

### Task 05 (b)

> Use doc.sents to split the paragraph into individual sentences.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 05 (b)
para = ("Dr. Smith met Mr. Jones at 5 p.m. on Monday to discuss the budget, e.g. salaries and travel. "
        "They agreed to meet again on Friday. Prof. Ali from the U.K. office joined later. "
        "Was the plan approved? Yes, it was!")

doc5 = nlp(para)
sentences = list(doc5.sents)
for i, sent in enumerate(sentences, 1):
    print(f"{i}. {sent.text}")
```

### Task 05 (c)

> Print the total number of sentences detected.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 05 (c)
print("Sentences detected:", len(sentences))
```

### Task 05 (d)

> Check whether the abbreviation caused any incorrect sentence break, and explain your observation in one line.

```python
# Zain Ul Wahaj | FA23-BAI-058 | Task 05 (d)
# The paragraph really has 5 sentences. A wrong break would split right after an abbreviation.
abbreviations = ["Dr.", "Mr.", "p.m.", "e.g.", "Prof.", "U.K."]
wrong_breaks = [a for a in abbreviations if any(s.text.strip().endswith(a) for s in sentences)]
print("Expected sentences          :", 5)
print("Detected sentences          :", len(sentences))
print("Sentences ending in an abbreviation (wrong breaks):", wrong_breaks or "none")

# OBSERVATION: no abbreviation caused a wrong break. spaCy's tokenizer keeps Dr., Mr., Prof., p.m.,
# e.g. and U.K. as single tokens, and the parser decides sentence starts from the whole sentence
# structure instead of just splitting at every full stop.
```
